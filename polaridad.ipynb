```python
from transformers import pipeline

# Cargar un pipeline de análisis de sentimientos en español
analizador = pipeline("sentiment-analysis", model="finiteautomata/beto-sentiment-analysis")

texto = "No me gustó para nada la atención, demoraron demasiado."
resultado = analizador(texto)[0]

print(f"Etiqueta: {resultado['label']}") # NEG, NEU, POS
print(f"Confianza/Intensidad: {resultado['score']:.2f}")
```

```python
from textblob import TextBlob

# Texto de ejemplo
texto = "¡Me encantó el servicio de entrega, el producto superó todas mis expectativas!"

# Análisis con TextBlob (calcula polaridad de -1.0 a 1.0)
analisis = TextBlob(texto)
polaridad = analisis.sentiment.polarity

print(f"Polaridad: {polaridad}")
if polaridad > 0:
  print("Sentimiento: Positivo")
elif polaridad < 0:
  print("Sentimiento: Negativo")
else:
  print("Sentimiento: Neutral")
```
