```python
import numpy as np

# Configurar señal (50 Hz y 120 Hz)
fs = 1000  # Frecuencia de muestreo (Hz)
t = np.linspace(0, 1.0, fs, endpoint=False)
senial = np.sin(2 * np.pi * 50 * t) + 0.5 * np.sin(2 * np.pi * 120 * t)

# Calcular FFT y vector de frecuencias
fft_valores = np.fft.fft(senial)
frecuencias = np.fft.fftfreq(len(senial), d=1/fs)

# Filtrar solo la mitad positiva del espectro
mitad = len(senial) // 2
frecuencias_pos = frecuencias[:mitad]
amplitudes_pos = (2.0 / len(senial)) * np.abs(fft_valores[:mitad])
```

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# Cargar un archivo de audio de ejemplo (reemplaza con tu archivo o usa uno de librosa)
audio_path = librosa.ex('nutcracker')
audio, sr = librosa.load(audio_path)

# ANÁLISIS DE ESPECTRO CON FFT
D = librosa.amplitude_to_db(np.abs(librosa.stft(audio)), ref=np.max)

plt.figure(figsize=(10, 4))
librosa.display.specshow(D, sr=sr, x_axis='time', y_axis='log', cmap='magma')
plt.colorbar(format='%+2.0f dB')
plt.title('Espectrograma de voz - FFT')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Crear la señal sintética (50 Hz + 120 Hz)
fs = 1000  # Frecuencia de muestreo (Hz)
t = np.linspace(0, 1.0, fs, endpoint=False)
senial = np.sin(2 * np.pi * 50 * t) + 0.7 * np.sin(2 * np.pi * 20 * t)

# 2. Calcular la FFT
fft_valores = np.fft.fft(senial)
frecuencias = np.fft.fftfreq(len(senial), d=1/fs)

# 3. Filtrar la mitad positiva del espectro
mitad = len(senial) // 2
frecuencias_pos = frecuencias[:mitad]
amplitudes_pos = (2.0 / len(senial)) * np.abs(fft_valores[:mitad])

# 4. Graficar Dominio del Tiempo y Dominio de la Frecuencia
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 6))

# Subplot 1: Señal en el tiempo (mostramos los primeros 0.1 segundos)
ax1.plot(t[:100], senial[:100], color='#1f77b4', linewidth=1.5)
ax1.set_title('Señal en el Dominio del Tiempo', fontsize=12, fontweight='bold')
ax1.set_xlabel('Tiempo [s]')
ax1.set_ylabel('Amplitud')
ax1.grid(True, linestyle='--', alpha=0.6)

# Subplot 2: Espectro de frecuencia
ax2.plot(frecuencias_pos, amplitudes_pos, color='#d62728', linewidth=1.5)
ax2.set_title('Espectro de Frecuencia (Transformada de Fourier)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Frecuencia [Hz]')
ax2.set_ylabel('Amplitud')
ax2.set_xlim(0, 200)  # Limitar hasta 200 Hz para ver claramente los picos
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
