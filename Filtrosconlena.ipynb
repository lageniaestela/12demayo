```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Descargar la imagen clásica de Lena directamente en el entorno de Colab
!wget -O lena.png https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg

# Leer la imagen y convertirla de BGR a RGB para visualizarla correctamente con Matplotlib
img_bgr = cv2.imread('/content/lena.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 1. Escala de grises y Desenfoque Gaussiano
gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (7, 7), 0)

# 2. Detección de Bordes (Canny)
edges = cv2.Canny(blurred, 100, 200)

# 3. Umbralización
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Mostrar los resultados en una sola figura de Matplotlib
plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1), plt.imshow(img_rgb), plt.title('Original (RGB)')
plt.axis('off')

plt.subplot(2, 3, 2), plt.imshow(gray, cmap='gray'), plt.title(
    'Escala de Grises'
)
plt.axis('off')

plt.subplot(2, 3, 3), plt.imshow(blurred, cmap='gray'), plt.title('Gaussian Blur')
plt.axis('off')

plt.subplot(2, 3, 4), plt.imshow(edges, cmap='gray'), plt.title(
    'Bordes (Canny)'
)
plt.axis('off')

plt.subplot(2, 3, 5), plt.imshow(thresh, cmap='gray'), plt.title(
    'Umbralización'
)
plt.axis('off')

plt.tight_layout()
plt.show()
```
